## Array Creation: Vectors, Matrices, Tensors

Quick ways to instantiate NumPy arrays for any shape or type.

```python
# Array creation: vectors, matrices, tensors
import numpy as np

a = np.array([1, 2, 3])  # 1D vector
A = np.array([[1, 2, 3], [4, 5, 6]])  # 2D matrix
Z = np.zeros((2, 3))  # zeros
O = np.ones((2, 3))   # ones
I = np.eye(3)         # identity
R = np.random.rand(2, 3)  # uniform random
N = np.random.randn(2, 3) # normal random
ar = np.arange(0, 6)      # range
lin = np.linspace(0, 1, 5) # linspace

print("a:", a)
print("A:\n", A)
print("I:\n", I)
```

## Vector Algebra: 1D Operations

Key vector ops: dot, outer, elementwise. 1D vs 2D behavior.

```python
# Vector ops: dot, outer, elementwise
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print("Dot:", v1 @ v2)
print("Outer:", np.outer(v1, v2))
print("Elementwise:", v1 * v2)
```

## Matrix Algebra: 2D Operations

Matrix multiplication, elementwise ops, and broadcasting essentials.

```python
# Matrix ops: multiplication, elementwise
C = np.array([[1, 2, 3], [4, 5, 6]])
D = np.array([[1, 2], [3, 4], [5, 6]])

print("C @ D:", C @ D)
print("Elementwise (C * D.T):", C * D.T)
```

## Shape Manipulation: 1D ↔ 2D Views

Efficiently convert vectors to row/column matrices for algebraic ops.

```python
# Shape manipulation: 1D ↔ 2D (3 ways)
v = np.array([1, 2, 3])
# Method 1: np.newaxis
row1 = v[np.newaxis, :]
col1 = v[:, np.newaxis]
# Method 2: reshape
row2 = v.reshape(1, -1)
col2 = v.reshape(-1, 1)
# Method 3: np.expand_dims
row3 = np.expand_dims(v, axis=0)
col3 = np.expand_dims(v, axis=1)
print('row:', row1, row1.shape)
print('col:', col1, col1.shape)
print('row (reshape):', row2, row2.shape)
print('col (reshape):', col2, col2.shape)
print('row (expand_dims):', row3, row3.shape)
print('col (expand_dims):', col3, col3.shape)
```
