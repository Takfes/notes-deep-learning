## Tensor Reshaping Methods


```python
# Start with a 1D tensor (500 data points)
x = torch.randn(500)
print(f"Original shape: {x.shape}")  # torch.Size([500])

# Method 1: reshape() - most flexible, creates new view if possible
x_reshaped = x.reshape(500, 1)  # or x.reshape(-1, 1)
print(f"reshape(500, 1): {x_reshaped.shape}")

# Method 2: view() - requires contiguous tensor, returns view
x_view = x.view(500, 1)  # faster but stricter requirements
print(f"view(500, 1): {x_view.shape}")

# Method 3: unsqueeze() - adds dimension of size 1
x_unsqueeze = x.unsqueeze(1)  # dimension 1 (column), or use unsqueeze(-1)
print(f"unsqueeze(1): {x_unsqueeze.shape}")

# Method 4: indexing with None/newaxis
x_index = x[:, None]  # or x[:, torch.newaxis]
print(f"[:, None]: {x_index.shape}")

# Key differences:
# - reshape(): most general, handles non-contiguous tensors
# - view(): fastest but requires contiguous memory layout
# - unsqueeze(): explicit about adding dimensions, most readable
# - indexing: numpy-style, good for quick dimension expansion

# Common neural network use case: batch dimension
x_batch = x.unsqueeze(0)  # add batch dimension at position 0
print(f"With batch dim: {x_batch.shape}")  # torch.Size([1, 500])
```

## NumPy Interoperability


```python
# NumPy to PyTorch
np_array = np.array([[1, 2], [3, 4]])
torch_from_np = torch.from_numpy(np_array)  # shares memory!
torch_copy = torch.tensor(np_array)  # creates copy

# PyTorch to NumPy (CPU tensors only)
torch_tensor = torch.randn(2, 3)
np_from_torch = torch_tensor.numpy()  # shares memory!
np_copy = torch_tensor.detach().numpy()  # for tensors with gradients

# Type conversions
float_np = np_array.astype(np.float32)
torch_float = torch.from_numpy(float_np)
print(f"NumPy dtype: {float_np.dtype}, PyTorch dtype: {torch_float.dtype}")

# Common workflow: numpy -> torch -> compute -> numpy
result = torch.from_numpy(np_array).float().matmul(torch.randn(2, 3)).numpy()
print(f"Computation result shape: {result.shape}")
```
