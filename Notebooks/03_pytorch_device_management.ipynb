# PyTorch Device Management - Quick Reference

## TL;DR
This notebook covers essential PyTorch device management:
- **Device Detection**: Check for CPU, CUDA (NVIDIA GPU), MPS (Apple Silicon)
- **Tensor Movement**: Move tensors between devices efficiently
- **Device Compatibility**: Ensure tensors are on same device for operations
- **Best Practices**: Automatic device selection and memory optimization

```python
import torch
import numpy as np

print(f"PyTorch version: {torch.__version__}")
```

## Device Detection

```python
# Check available devices
cuda_available = torch.cuda.is_available()
mps_available = torch.backends.mps.is_available()

print(f"CUDA: {cuda_available}")
print(f"MPS: {mps_available}")

# Automatic device selection
def get_device():
    if torch.cuda.is_available():
        return torch.device("cuda")
    elif torch.backends.mps.is_available():
        return torch.device("mps")
    else:
        return torch.device("cpu")

device = get_device()
print(f"Best device: {device}")
```

## Moving Tensors Between Devices

```python
# Create tensor on CPU (default)
x = torch.randn(3, 3)

# Movement methods
x_cpu = x.to('cpu')          # string device
x_cpu = x.to(device)         # device object
x_cpu = x.cpu()              # explicit method

# Move to GPU (if available)
if torch.cuda.is_available():
    x_cuda = x.to('cuda')     # or x.cuda()
    x_cuda = x.to('cuda:0')   # specific GPU
    
if torch.backends.mps.is_available():
    x_mps = x.to('mps')

# Create directly on device (preferred)
x_direct = torch.randn(3, 3, device=device)
print(f"Created on: {x_direct.device}")
```

## Device Compatibility

```python
# Tensors must be on same device for operations
a = torch.randn(2, 2)      # CPU
b = torch.randn(2, 2)      # CPU
result = a @ b             # ✅ Works - both on CPU

# Device mismatch example
if device.type != 'cpu':
    a_gpu = a.to(device)   # Move to GPU/MPS
    # result = a + a_gpu   # ❌ Would fail - different devices
    
    # Solutions:
    result1 = a.to(device) + a_gpu    # Move a to device
    result2 = a + a_gpu.cpu()         # Move a_gpu to CPU
    
    print(f"Compatible operations: {result1.device}, {result2.device}")
```
